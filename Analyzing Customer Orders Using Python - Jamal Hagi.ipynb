
#### Task #1 Store customer orders 

```python
customer_name = ['Alice' , 'John' , 'Mark' , 'Ray' , 'Bob' , 'Jane' , 'Mike'] # cutomer names list

# Entering order details tuples which contain customer name, product, price , category

order_details = [
    ('Alice' , 'Airpod' , 200 , 'electronics' ),
    ('Alice' , 'Airpod case' , 25 , 'electronics'),
    ('Alice' , 'shirt' , 25 , 'clothing'),
    ('John' , 'apple' , 5 , 'food'),
    ('John' , 'rice' , 3.8, 'food'),
    ('John' , 'sugar' , 2.53, 'food'),
    ('John' , 'screw driver', 3.0 , 'tools'),
    ('John' , 'tupperware' , 20 , 'kitchen'),
    ('Mark' , 'paint' , 80 , 'paints'),
    ('Mark' , 'brush' , 11.2 , 'paints'),
    ('Mark' , 'gloves' , 3.5 ,'paints'),
    ('Ray' , 'watch' , 250 , 'electronics'),
    ('Ray' , 'pliers' , 20 , 'tools'),
    ('Ray' , 'ice cream' , 6 , 'food'),
    ('Bob' , 'meat' , 12 , 'food'),
    ('Bob' , 'bowl' , 16 , 'kitchen'),
    ('Bob' , 'cups' , 8 , 'kitchen'),
    ('Jane' , 'pens' , 6 , 'office'),
    ('Jane' , 'paper' , 8 , 'office'),
    ('Jane' , 'printer' , 300 , 'office'),
    ('Mike' , 'pants' , 25 , 'clothing'),
    ('Mike' , 'headphone' , 13 , 'electronics'),


]
print ('Here is the customer name list:')
print (customer_name)

#Creating dictionaty to map customer name to ordered products
 
orders_dict = {}
for order in order_details:
    orders_dict.setdefault(order[0], []).append(order[1])

print ('\nHere is the Order Dictionary:')
print (orders_dict)
```

#### Task #2  Classify products by category

```python
# using dictionary to map product to its respoective category

product_category_dict = {}
for order in order_details:
    product_category_dict[order[1]] = order[3]
    
print ('Here is the Product Category Dictionary:')
print (product_category_dict)

# creating a set of unique product category
category_set = set(product_category_dict.values())

print ('Here is all Unique Product Category set:')
print(category_set)
```

#### Task #3   Analyze customer orders 

```python
#Use a loop to calculate the total amount each customer spent

print("\nHere is the total spend per customer:")

total_spend_dict = {}

for n in order_details:
    customer = n[0]
    price = n[2]
    total_spend_dict.setdefault(customer,0) #create dictionary for each entry (Customer, spending) and assumes zero initial spending
    total_spend_dict[customer] += price

for customer, total in total_spend_dict.items():
    print (f"{customer}: {total}")


# If the total purchase value is above $100, classify the customer as a high-value buyer

print ("\nHere is the customer clasification")
for customer, total in total_spend_dict.items():
    if total > 100 :
        print(f"{customer}:'high-value buyer '")
    elif 50 <= total <= 100:
        print(f"{customer}:'moderate buyer '")
    else:
        print(f"{customer}:'low-value buyer '")
```

#### Task #4   Generate business insights 

```python
# Calculating the total revenue per product category and store it in a dictionary

print ("\nHere is the total revenue per category ")

total_rev_dict = {}

for k in order_details:
    category = k[3]
    price = k[2]
    total_rev_dict.setdefault(category,0) #create dictionary for each entry (category) and assumes zero initial spending
    total_rev_dict[category] += price

for category, rev in total_rev_dict.items():
    print (f"{category}: {rev}")

# Extracting unique products from all orders
unique_products = {order[1] for order in order_details}

print("\nHere are the unique products ordered:")
print(unique_products)


# Using a list comprehension to find all customers who purchased electronics
print("\n Here is a list of customer who bought electronics")
electronics_customers = list({order[0] for order in order_details if order[3] == 'electronics'})
print(electronics_customers)


# Identifying the top three highest-spending customers using sorting 
print('\ntotal_spend_dict')
print (total_spend_dict)
top_three_dic= sorted(total_spend_dict.items(), key=lambda x :x [1],reverse=True)[:3]
print('\ntop three dict')
print(top_three_dic)
top_three_cust_names = dict(top_three_dic).keys()
print('\ntop three customer names')
print(list(top_three_cust_names))
```

#### Task #5   Organize and display data

```python
# Printing a summary of each customer’s total spending and their classification 

print("\ncustomer’s total spending & classification")
print("Customer   | Total Spend | Classification")
print("-----------------------------------------")
for customer, total in total_spend_dict.items():
    if total > 100:
        classification = "high-value buyer"
    elif 50 <= total <= 100:
        classification = "moderate buyer"
    else:
        classification = "low-value buyer"
    print(f"{customer:<10} | {total:<11.2f} | {classification}")

#Using set operations to find customers who purchased from multiple categories

print('\nList of customer bought from multiple categories:')
customer_categories_dict = {}
for x in order_details:
    customer = x[0]
    category = x[3]
    customer_categories_dict.setdefault(customer, []).append(category)

for customer, categories in customer_categories_dict.items():
    if len(set(categories)) > 1:
        print(customer)

# Finding customers who bought both electronics and clothing
print("\nCustomers who bought both electronics and clothing:")
for customer, categories in customer_categories_dict.items():
    categories_set = set(categories)  # convert list to set to remove duplicates
    if 'electronics' in categories_set and 'clothing' in categories_set:
        print(customer)
```
